```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format","{:.2f}".format)

plt.rcParams["figure.figsize"] = (8,5)
```

```python
df = pd.read_csv("../data/processed_data/cleaned_dataset.csv")

print("Dataset Loaded Successfully!")
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df,x="Heart_Disease")

plt.title("Heart Disease Distribution")
plt.xlabel("Heart Disease")
plt.ylabel("Count")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(14,10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df[["age","BMI","Systolic_BP","Diastolic_BP"]])

plt.title("Boxplot")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Heart_Disease",
    y="age",
    data=df
)

plt.title("Heart Disease vs Age")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Heart_Disease",
    y="BMI",
    data=df
)

plt.title("Heart Disease vs BMI")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Heart_Disease",
    y="Systolic_BP",
    data=df
)

plt.title("Heart Disease vs Systolic BP")

plt.show()
```

```python
pair_columns = [
    "age",
    "BMI",
    "Systolic_BP",
    "Diastolic_BP",
    "Heart_Disease"
]

sns.pairplot(
    df[pair_columns],
    hue="Heart_Disease"
)

plt.show()
```

```python
corr = numeric_df.corr()

corr.to_csv("../outputs/correlation_matrix.csv")

print("Correlation Matrix Saved!")
```
