```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

pd.set_option("display.max_columns",None)
```

```python
df = pd.read_csv("../data/processed_data/cleaned_dataset.csv")

print("Dataset Loaded Successfully")
```

```python
print(df.columns)
```

```python
# Remove unnecessary and redundant columns

df.drop(
    columns=[
        "id",                 # Identifier
        "Height_m",           # Temporary column used only for BMI calculation
        "Age_Group",          # Derived from age
        "BMI_Category",       # Derived from BMI
        "BP_Category"         # Derived from blood pressure
    ],
    inplace=True
)

print("Remaining Columns:\n")
print(df.columns)

print("\nTotal Columns:", len(df.columns))
```

```python
X = df.drop("Heart_Disease",axis=1)

y = df["Heart_Disease"]
```

```python
print("Features :",X.shape)

print("Target :",y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)
```

```python
print("Training Features :",X_train.shape)

print("Testing Features :",X_test.shape)

print("Training Labels :",y_train.shape)

print("Testing Labels :",y_test.shape)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
print(X_train[:5])
```

```python
X_train = pd.DataFrame(
    X_train,
    columns=X.columns
)

X_test = pd.DataFrame(
    X_test,
    columns=X.columns
)
```

```python
feature_dataset = df.copy()

feature_dataset.to_csv(

    "../data/processed_data/feature_engineered_dataset.csv",

    index=False

)

print("Feature Engineered Dataset Saved")
```

```python
X_train.to_csv(

    "../data/train_test_data/X_train.csv",

    index=False

)

X_test.to_csv(

    "../data/train_test_data/X_test.csv",

    index=False

)

y_train.to_csv(

    "../data/train_test_data/y_train.csv",

    index=False

)

y_test.to_csv(

    "../data/train_test_data/y_test.csv",

    index=False

)

print("Train Test Dataset Saved")
```
